```python
class PadeActivation(nn.Module):
    """
    Learnable Padé rational activation:

        P(x)
    y = -----
        Q(x)

    numerator: degree 3
    denominator: degree 2
    """

    def __init__(
        self,
        channels,
        m=3,
        n=2
    ):
        super().__init__()

        self.m = m
        self.n = n

        # 每个 channel 独立的 Padé 系数
        self.a = nn.Parameter(
            torch.zeros(
                channels,
                m + 1
            )
        )

        self.b = nn.Parameter(
            torch.zeros(
                channels,
                n
            )
        )

        # 初始接近 identity
        with torch.no_grad():

            self.a[:, 0] = 0.0
            self.a[:, 1] = 1.0

            if m >= 2:
                self.a[:, 2:] = 0.0

            self.b[:] = 0.0

    def forward(self, x):
        # x: [B,C,L]

        # 防止高阶项数值爆炸
        z = torch.clamp(
            x,
            -5.0,
            5.0
        )

        # ----------------------------
        # numerator
        # ----------------------------

        numerator = 0.0

        for i in range(
            self.m + 1
        ):

            coefficient = self.a[
                :, i
            ].view(
                1, -1, 1
            )

            numerator = (
                numerator
                + coefficient
                * z.pow(i)
            )

        # ----------------------------
        # denominator
        # ----------------------------

        denominator_poly = 0.0

        for j in range(
            1,
            self.n + 1
        ):

            coefficient = self.b[
                :, j - 1
            ].view(
                1, -1, 1
            )

            denominator_poly = (
                denominator_poly
                + coefficient
                * z.pow(j)
            )

        denominator = (
            1.0
            + torch.abs(
                denominator_poly
            )
        )

        return (
            numerator
            / denominator
        )
```

```python
class PadeFeatureExtractor(nn.Module):

    def __init__(
        self,
        in_channels=2,
        hidden_channels=16,
        dropout=0.1
    ):
        super().__init__()

        self.input_conv = nn.Conv1d(
            in_channels,
            hidden_channels,
            kernel_size=7,
            padding=3
        )

        self.pade1 = PadeActivation(
            hidden_channels,
            m=3,
            n=2
        )

        self.depthwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=5,
            padding=2,
            groups=hidden_channels
        )

        self.pointwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=1
        )

        self.pade2 = PadeActivation(
            hidden_channels,
            m=3,
            n=2
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.output_conv = nn.Conv1d(
            hidden_channels,
            in_channels,
            kernel_size=1
        )

        # 防止 Padé 一开始破坏原始信号
        self.scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):

        # [B,L,2]
        residual = x

        # -> [B,2,L]
        y = x.transpose(
            1,
            2
        )

        y = self.input_conv(y)

        y = self.pade1(y)

        y = self.depthwise(y)
        y = self.pointwise(y)

        y = self.pade2(y)

        y = self.dropout(y)

        y = self.output_conv(y)

        # -> [B,L,2]
        y = y.transpose(
            1,
            2
        )

        return (
            residual
            + self.scale * y
        )
```

```python
class PatchEmbedding(nn.Module):
    def __init__(
        self,
        in_ch=2,
        patch_len=64,
        stride=64,
        d_model=64
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        self.proj = nn.Linear(
            in_ch * patch_len,
            d_model
        )

    def forward(self, x):
        # x: [B, L, C]

        p = x.unfold(
            dimension=1,
            size=self.patch_len,
            step=self.stride
        )

        # [B, N, C, P] -> [B, N, C*P]
        p = p.contiguous().view(
            x.size(0),
            p.size(1),
            -1
        )

        return self.proj(p)


print("PatchEmbedding 定义成功")
```

```python
class SmallGDN2(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads=4,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.dh = d_model // n_heads

        self.qkv = nn.Linear(
            d_model,
            3 * d_model,
            bias=False
        )

        self.erase_gate = nn.Linear(
            d_model,
            d_model
        )

        self.write_gate = nn.Linear(
            d_model,
            d_model
        )

        self.decay_gate = nn.Linear(
            d_model,
            n_heads
        )

        self.out_gate = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.norm = nn.LayerNorm(
            self.dh
        )

        self.dropout = nn.Dropout(
            dropout
        )

        # ==========================================
        # 新增 1：Gate initialization
        # ==========================================

        # 初始时更倾向保留历史状态
        nn.init.constant_(
            self.decay_gate.bias,
            2.0
        )

        # 初始 erase 不要太强
        nn.init.constant_(
            self.erase_gate.bias,
            -1.0
        )

        # 初始 write 也不要过强
        nn.init.constant_(
            self.write_gate.bias,
            -1.0
        )

        # ==========================================
        # 新增 2：可学习 residual scale
        # ==========================================

        self.res_scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):

        B, N, D = x.shape

        H = self.n_heads
        dh = self.dh

        q, k, v = self.qkv(x).chunk(
            3,
            dim=-1
        )

        q = q.view(
            B, N, H, dh
        ).transpose(1, 2)

        k = k.view(
            B, N, H, dh
        ).transpose(1, 2)

        v = v.view(
            B, N, H, dh
        ).transpose(1, 2)

        q = F.normalize(
            q,
            dim=-1
        )

        k = F.normalize(
            k,
            dim=-1
        )

        erase = torch.sigmoid(
            self.erase_gate(x)
        )

        erase = erase.view(
            B, N, H, dh
        ).transpose(1, 2)

        write = torch.sigmoid(
            self.write_gate(x)
        )

        write = write.view(
            B, N, H, dh
        ).transpose(1, 2)

        decay = torch.sigmoid(
            self.decay_gate(x)
        ).transpose(1, 2)

        state = x.new_zeros(
            B,
            H,
            dh,
            dh
        )

        outputs = []

        for t in range(N):

            qt = q[:, :, t]
            kt = k[:, :, t]
            vt = v[:, :, t]

            et = erase[:, :, t]
            wt = write[:, :, t]

            dt = decay[
                :, :, t
            ].unsqueeze(
                -1
            ).unsqueeze(-1)

            # decay
            state = dt * state

            # erase
            erase_vector = torch.einsum(
                "bhd,bhdv->bhv",
                et * kt,
                state
            )

            state = (
                state
                - kt.unsqueeze(-1)
                * erase_vector.unsqueeze(-2)
            )

            # write
            state = (
                state
                + kt.unsqueeze(-1)
                * (wt * vt).unsqueeze(-2)
            )

            # read
            yt = torch.einsum(
                "bhd,bhdv->bhv",
                qt,
                state
            )

            outputs.append(yt)

        y = torch.stack(
            outputs,
            dim=2
        )

        y = self.norm(y)

        y = (
            y.transpose(1, 2)
            .contiguous()
            .view(B, N, D)
        )

        gate = torch.sigmoid(
            self.out_gate(x)
        )

        y = self.out_proj(
            y * gate
        )

        y = self.dropout(y)

        # 核心新增
        y = self.res_scale * y

        return y, None
```

```python
class TFPSBearing(nn.Module):

    def __init__(
        self,
        mixer="mha",
        use_pade=False,          # 新增
        seq_len=2048,
        patch_len=64,
        stride=64,
        d_model=64,
        n_heads=4,
        n_experts=4,
        top_k=2,
        d_ff=128,
        dropout=0.1
    ):
        super().__init__()

        self.mixer_name = mixer
        self.use_pade = use_pade

        # =========================================
        # 新增：Padé 特征提取
        # =========================================
        if use_pade:
            self.pade_feature = PadeFeatureExtractor(
                in_channels=2,
                hidden_channels=16,
                dropout=dropout
            )

        # Patch Embedding
        self.patch = PatchEmbedding(
            in_ch=2,
            patch_len=patch_len,
            stride=stride,
            d_model=d_model
        )

        self.pre_norm = nn.LayerNorm(d_model)

        # =========================================
        # 时域分支
        # =========================================
        if mixer == "mha":

            self.time_mixer = MHATokenMixer(
                d_model,
                n_heads,
                dropout
            )

        elif mixer == "gdn2":

            self.time_mixer = SmallGDN2(
                d_model,
                n_heads,
                dropout
            )

        else:
            raise ValueError(
                f"Unknown mixer: {mixer}"
            )

        self.time_ffn = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, 2 * d_model),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(2 * d_model, d_model)
        )

        # =========================================
        # 频域分支
        # =========================================
        self.freq = FrequencyBranch(
            d_model
        )

        # =========================================
        # 聚类 / Router
        # =========================================
        self.router = PrototypeRouter(
            d_model,
            n_experts,
            top_k
        )

        # =========================================
        # MoE
        # =========================================
        self.moe = PatternMoE(
            d_model,
            d_ff,
            n_experts
        )

        # =========================================
        # 融合
        # =========================================
        self.fuse_gate = nn.Sequential(
            nn.Linear(
                2 * d_model,
                d_model
            ),
            nn.Sigmoid()
        )

        # =========================================
        # RUL prediction head
        # =========================================
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                d_model // 2
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                d_model // 2,
                1
            )
        )

    # =====================================================
    # 这里就是你之前找不到的 forward
    # =====================================================
    def forward(self, x):

        # x: [B, 2048, 2]

        # =========================================
        # 新增：Padé 特征提取
        # =========================================
        if self.use_pade:
            x = self.pade_feature(x)

        # =========================================
        # Patch embedding
        # =========================================
        tokens = self.patch(x)

        tokens = self.pre_norm(
            tokens
        )

        # =========================================
        # Time branch
        # =========================================
        time_mix, attention = (
            self.time_mixer(tokens)
        )

        time_feature = (
            tokens + time_mix
        )

        time_feature = (
            time_feature
            + self.time_ffn(
                time_feature
            )
        )

        # =========================================
        # Frequency branch
        # =========================================
        freq_feature = self.freq(
            tokens
        )

        # =========================================
        # Clustering / Router
        # =========================================
        common = 0.5 * (
            time_feature
            + freq_feature
        )

        router_weights, balance_loss = (
            self.router(common)
        )

        # =========================================
        # MoE
        # =========================================
        moe_feature = self.moe(
            common,
            router_weights
        )

        # =========================================
        # Fusion
        # =========================================
        gate = self.fuse_gate(
            torch.cat(
                [
                    time_feature,
                    freq_feature
                ],
                dim=-1
            )
        )

        fused = (
            gate * time_feature
            + (1.0 - gate)
            * freq_feature
        )

        fused = (
            fused + moe_feature
        )

        # =========================================
        # RUL prediction
        # =========================================
        pooled = fused.mean(
            dim=1
        )

        pred = self.head(
            pooled
        ).squeeze(-1)

        return pred, {
            "balance_loss": balance_loss,
            "router": router_weights,
            "attention": attention
        }


print("新的 TFPSBearing 已定义")
```

```python
def preprocess_signal(horizontal, vertical, seq_len=2048):

    h = np.asarray(horizontal, dtype=np.float32)
    v = np.asarray(vertical, dtype=np.float32)

    n = min(len(h), len(v))

    idx = np.linspace(
        0,
        n - 1,
        seq_len
    ).astype(np.int64)

    x = np.stack(
        [h[idx], v[idx]],
        axis=-1
    )

    # 注意：这里不再单独标准化每个 snapshot
    return x.astype(np.float32)

def build_xjtu_cache(ds, cache_path):

    meta = ds.metadata()

    # horizontal / vertical 在 metadata 中可能各占一行
    records = (
        meta
        .sort_values(
            ["run_id", "snapshot", "recording_id"]
        )
        .drop_duplicates("recording_id")
        .reset_index(drop=True)
    )

    # 每个轴承最大剩余寿命
    max_rul_dict = (
        records
        .groupby("run_id")["rul_s"]
        .max()
        .to_dict()
    )

    X = []
    Y = []

    bearing_number = []
    bearing_name = []
    snapshot_id = []

    for row in tqdm(
        records.itertuples(index=False),
        total=len(records)
    ):

        signals = ds.recording(
            row.recording_id,
            channels=[
                "horizontal",
                "vertical"
            ]
        )

        x = preprocess_signal(
            signals["horizontal"],
            signals["vertical"],
            SEQ_LEN
        )

        run_id = str(row.run_id)

        # Bearing1_1 -> 最后的 1
        m = re.fullmatch(
            r"Bearing(\d+)_(\d+)",
            run_id
        )

        if m is None:
            continue

        number = int(m.group(2))

        max_rul = max_rul_dict[run_id]

        if max_rul > 0:
            rul = float(row.rul_s) / float(max_rul)
        else:
            rul = 0.0

        X.append(x)
        Y.append(rul)

        bearing_number.append(number)
        bearing_name.append(run_id)
        snapshot_id.append(int(row.snapshot))

    X = np.stack(X).astype(np.float32)
    Y = np.asarray(Y, dtype=np.float32)

    bearing_number = np.asarray(
        bearing_number,
        dtype=np.int64
    )

    bearing_name = np.asarray(bearing_name)

    snapshot_id = np.asarray(
        snapshot_id,
        dtype=np.int64
    )

    np.savez_compressed(
        cache_path,
        X=X,
        y=Y,
        fold=bearing_number,
        key=bearing_name,
        sid=snapshot_id
    )

    return (
        X,
        Y,
        bearing_number,
        bearing_name,
        snapshot_id
    )


if CACHE_PATH.exists():

    print("读取缓存")

    z = np.load(
        CACHE_PATH,
        allow_pickle=True
    )

    X = z["X"]
    y = z["y"]
    fold = z["fold"]
    key = z["key"]
    sid = z["sid"]

else:

    X, y, fold, key, sid = build_xjtu_cache(
        ds,
        CACHE_PATH
    )


print("X:", X.shape)
print("y:", y.shape)

print(
    "RUL range:",
    y.min(),
    y.max()
)
```

```python
class BearingRULDataset(Dataset):

    def __init__(self, X, y, mask):
        self.X = X[mask]
        self.y = y[mask]

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):

        x = torch.from_numpy(
            self.X[idx]
        )

        y = torch.tensor(
            self.y[idx],
            dtype=torch.float32
        )

        return x, y


# 每种工况：
# *_1 *_2 *_3 -> train
# *_4          -> validation
# *_5          -> test

train_mask = np.isin(
    fold,
    [1, 2, 3]
)

val_mask = fold == 4
test_mask = fold == 5

# =====================================================
# 只用训练集计算全局 mean/std
# 避免验证/测试信息泄漏
# =====================================================

train_mean = X[train_mask].mean(
    axis=(0, 1),
    keepdims=True
)

train_std = X[train_mask].std(
    axis=(0, 1),
    keepdims=True
) + 1e-6

X = (
    X - train_mean
) / train_std

X = X.astype(np.float32)

print("train mean:", train_mean)
print("train std:", train_std)

train_ds = BearingRULDataset(
    X,
    y,
    train_mask
)

val_ds = BearingRULDataset(
    X,
    y,
    val_mask
)

test_ds = BearingRULDataset(
    X,
    y,
    test_mask
)


train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)


print(
    "train:",
    len(train_ds)
)

print(
    "validation:",
    len(val_ds)
)

print(
    "test:",
    len(test_ds)
)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# =========================================================
# 1. Padé activation
# =========================================================
class PadeActivation(nn.Module):
    def __init__(self, channels, m=3, n=2):
        super().__init__()

        self.m = m
        self.n = n

        self.a = nn.Parameter(
            torch.zeros(channels, m + 1)
        )

        self.b = nn.Parameter(
            torch.zeros(channels, n)
        )

        # 初始近似 identity
        with torch.no_grad():
            self.a[:, 0] = 0.0
            self.a[:, 1] = 1.0

            if m >= 2:
                self.a[:, 2:] = 0.0

            self.b[:] = 0.0

    def forward(self, x):
        # x: [B,C,L]

        z = torch.clamp(x, -5.0, 5.0)

        numerator = torch.zeros_like(z)

        for i in range(self.m + 1):
            coef = self.a[:, i].view(1, -1, 1)
            numerator = numerator + coef * z.pow(i)

        denominator_poly = torch.zeros_like(z)

        for j in range(1, self.n + 1):
            coef = self.b[:, j - 1].view(1, -1, 1)
            denominator_poly = (
                denominator_poly
                + coef * z.pow(j)
            )

        denominator = (
            1.0
            + torch.abs(denominator_poly)
        )

        return numerator / denominator


# =========================================================
# 2. Padé feature extractor
# =========================================================
class PadeFeatureExtractor(nn.Module):
    def __init__(
        self,
        in_channels=2,
        hidden_channels=16,
        dropout=0.1
    ):
        super().__init__()

        self.input_conv = nn.Conv1d(
            in_channels,
            hidden_channels,
            kernel_size=7,
            padding=3
        )

        self.pade1 = PadeActivation(
            hidden_channels
        )

        self.depthwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=5,
            padding=2,
            groups=hidden_channels
        )

        self.pointwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=1
        )

        self.pade2 = PadeActivation(
            hidden_channels
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.output_conv = nn.Conv1d(
            hidden_channels,
            in_channels,
            kernel_size=1
        )

        self.scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):
        # x: [B,L,2]

        residual = x

        y = x.transpose(1, 2)

        y = self.input_conv(y)
        y = self.pade1(y)

        y = self.depthwise(y)
        y = self.pointwise(y)

        y = self.pade2(y)
        y = self.dropout(y)

        y = self.output_conv(y)

        y = y.transpose(1, 2)

        return residual + self.scale * y


# =========================================================
# 3. PatchEmbedding
# =========================================================
class PatchEmbedding(nn.Module):
    def __init__(
        self,
        in_ch=2,
        patch_len=64,
        stride=64,
        d_model=64
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        self.proj = nn.Linear(
            in_ch * patch_len,
            d_model
        )

    def forward(self, x):
        # [B,L,C]

        p = x.unfold(
            dimension=1,
            size=self.patch_len,
            step=self.stride
        )

        p = p.contiguous().view(
            x.size(0),
            p.size(1),
            -1
        )

        return self.proj(p)


# =========================================================
# 4. Baseline MHA
# =========================================================
class MHATokenMixer(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        dropout=0.1
    ):
        super().__init__()

        self.attn = nn.MultiheadAttention(
            d_model,
            n_heads,
            dropout=dropout,
            batch_first=True
        )

    def forward(self, x):

        y, attn = self.attn(
            x,
            x,
            x,
            need_weights=False
        )

        return y, attn


# =========================================================
# 5. Small GDN-2
# =========================================================
class SmallGDN2(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads=4,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.dh = d_model // n_heads

        self.qkv = nn.Linear(
            d_model,
            3 * d_model,
            bias=False
        )

        self.erase_gate = nn.Linear(
            d_model,
            d_model
        )

        self.write_gate = nn.Linear(
            d_model,
            d_model
        )

        self.decay_gate = nn.Linear(
            d_model,
            n_heads
        )

        self.out_gate = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.norm = nn.LayerNorm(
            self.dh
        )

        self.dropout = nn.Dropout(
            dropout
        )

        # gate initialization
        nn.init.constant_(
            self.decay_gate.bias,
            2.0
        )

        nn.init.constant_(
            self.erase_gate.bias,
            -1.0
        )

        nn.init.constant_(
            self.write_gate.bias,
            -1.0
        )

        # learnable residual scale
        self.res_scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):

        B, N, D = x.shape

        H = self.n_heads
        dh = self.dh

        q, k, v = self.qkv(x).chunk(
            3,
            dim=-1
        )

        q = q.view(
            B, N, H, dh
        ).transpose(1, 2)

        k = k.view(
            B, N, H, dh
        ).transpose(1, 2)

        v = v.view(
            B, N, H, dh
        ).transpose(1, 2)

        q = F.normalize(q, dim=-1)
        k = F.normalize(k, dim=-1)

        erase = torch.sigmoid(
            self.erase_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        write = torch.sigmoid(
            self.write_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        decay = torch.sigmoid(
            self.decay_gate(x)
        ).transpose(1, 2)

        state = x.new_zeros(
            B,
            H,
            dh,
            dh
        )

        outputs = []

        for t in range(N):

            qt = q[:, :, t]
            kt = k[:, :, t]
            vt = v[:, :, t]

            et = erase[:, :, t]
            wt = write[:, :, t]

            dt = decay[
                :, :, t
            ].unsqueeze(-1).unsqueeze(-1)

            # decay
            state = dt * state

            # erase
            erase_vec = torch.einsum(
                "bhd,bhdv->bhv",
                et * kt,
                state
            )

            state = (
                state
                - kt.unsqueeze(-1)
                * erase_vec.unsqueeze(-2)
            )

            # write
            state = (
                state
                + kt.unsqueeze(-1)
                * (wt * vt).unsqueeze(-2)
            )

            # read
            yt = torch.einsum(
                "bhd,bhdv->bhv",
                qt,
                state
            )

            outputs.append(yt)

        y = torch.stack(
            outputs,
            dim=2
        )

        y = self.norm(y)

        y = (
            y.transpose(1, 2)
            .contiguous()
            .view(B, N, D)
        )

        gate = torch.sigmoid(
            self.out_gate(x)
        )

        y = self.out_proj(
            y * gate
        )

        y = self.dropout(y)

        y = self.res_scale * y

        return y, None


# =========================================================
# 6. Frequency branch
# =========================================================
class FrequencyBranch(nn.Module):
    def __init__(self, d_model):
        super().__init__()

        self.proj = nn.Linear(
            2 * d_model,
            d_model
        )

        self.ffn = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(
                d_model,
                2 * d_model
            ),
            nn.GELU(),
            nn.Linear(
                2 * d_model,
                d_model
            )
        )

    def forward(self, tokens):

        z = torch.fft.fft(
            tokens,
            dim=1,
            norm="ortho"
        )

        z = torch.cat(
            [
                z.real,
                z.imag
            ],
            dim=-1
        )

        z = self.proj(z)

        return z + self.ffn(z)


# =========================================================
# 7. Router
# =========================================================
class PrototypeRouter(nn.Module):
    def __init__(
        self,
        d_model,
        n_experts=4,
        top_k=2,
        temperature=0.5
    ):
        super().__init__()

        self.n_experts = n_experts
        self.top_k = top_k
        self.temperature = temperature

        self.prototypes = nn.Parameter(
            torch.randn(
                n_experts,
                d_model
            ) * 0.02
        )

    def forward(self, x):

        # 用 float32 避免 AMP overflow
        xf = x.float()
        pf = self.prototypes.float()

        xn = F.normalize(
            xf,
            dim=-1
        )

        pn = F.normalize(
            pf,
            dim=-1
        )

        logits = torch.einsum(
            "bnd,kd->bnk",
            xn,
            pn
        )

        logits = (
            logits
            / self.temperature
        )

        if self.top_k < self.n_experts:

            values, indices = torch.topk(
                logits,
                self.top_k,
                dim=-1
            )

            sparse = torch.full_like(
                logits,
                torch.finfo(
                    logits.dtype
                ).min
            )

            sparse.scatter_(
                -1,
                indices,
                values
            )

            weights = torch.softmax(
                sparse,
                dim=-1
            )

        else:

            weights = torch.softmax(
                logits,
                dim=-1
            )

        usage = weights.mean(
            dim=(0, 1)
        )

        balance_loss = (
            (
                usage
                - 1.0 / self.n_experts
            ) ** 2
        ).mean()

        weights = weights.to(
            dtype=x.dtype
        )

        return weights, balance_loss


# =========================================================
# 8. MoE
# =========================================================
class PatternMoE(nn.Module):
    def __init__(
        self,
        d_model,
        d_ff,
        n_experts
    ):
        super().__init__()

        self.experts = nn.ModuleList([
            nn.Sequential(
                nn.LayerNorm(d_model),

                nn.Linear(
                    d_model,
                    d_ff
                ),

                nn.GELU(),

                nn.Linear(
                    d_ff,
                    d_model
                )
            )
            for _ in range(n_experts)
        ])

    def forward(
        self,
        x,
        weights
    ):

        outputs = torch.stack(
            [
                expert(x)
                for expert in self.experts
            ],
            dim=-2
        )

        return (
            outputs
            * weights.unsqueeze(-1)
        ).sum(dim=-2)


# =========================================================
# 9. 完整 TFPSBearing
# =========================================================
class TFPSBearing(nn.Module):
    def __init__(
        self,
        mixer="mha",
        use_pade=False,
        seq_len=2048,
        patch_len=64,
        stride=64,
        d_model=64,
        n_heads=4,
        n_experts=4,
        top_k=2,
        d_ff=128,
        dropout=0.1
    ):
        super().__init__()

        self.mixer_name = mixer
        self.use_pade = use_pade

        # Padé
        if use_pade:
            self.pade_feature = (
                PadeFeatureExtractor(
                    in_channels=2,
                    hidden_channels=16,
                    dropout=dropout
                )
            )

        # patch
        self.patch = PatchEmbedding(
            in_ch=2,
            patch_len=patch_len,
            stride=stride,
            d_model=d_model
        )

        self.pre_norm = nn.LayerNorm(
            d_model
        )

        # time branch
        if mixer == "mha":

            self.time_mixer = (
                MHATokenMixer(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        elif mixer == "gdn2":

            self.time_mixer = (
                SmallGDN2(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        else:

            raise ValueError(
                f"Unknown mixer: {mixer}"
            )

        self.time_ffn = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                2 * d_model
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                2 * d_model,
                d_model
            )
        )

        # frequency
        self.freq = FrequencyBranch(
            d_model
        )

        # router
        self.router = PrototypeRouter(
            d_model,
            n_experts,
            top_k
        )

        # MoE
        self.moe = PatternMoE(
            d_model,
            d_ff,
            n_experts
        )

        # fusion
        self.fuse_gate = nn.Sequential(
            nn.Linear(
                2 * d_model,
                d_model
            ),
            nn.Sigmoid()
        )

        # prediction head
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                d_model // 2
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                d_model // 2,
                1
            )
        )

    def forward(self, x):

        # ===============================
        # Padé
        # ===============================
        if self.use_pade:
            x = self.pade_feature(x)

        # ===============================
        # Patch
        # ===============================
        tokens = self.patch(x)

        tokens = self.pre_norm(
            tokens
        )

        # ===============================
        # Time branch
        # ===============================
        time_mix, attention = (
            self.time_mixer(tokens)
        )

        time_feature = (
            tokens + time_mix
        )

        time_feature = (
            time_feature
            + self.time_ffn(
                time_feature
            )
        )

        # ===============================
        # Frequency branch
        # ===============================
        freq_feature = self.freq(
            tokens
        )

        # ===============================
        # Router
        # ===============================
        common = 0.5 * (
            time_feature
            + freq_feature
        )

        router_weights, balance_loss = (
            self.router(common)
        )

        # ===============================
        # MoE
        # ===============================
        moe_feature = self.moe(
            common,
            router_weights
        )

        # ===============================
        # Fusion
        # ===============================
        gate = self.fuse_gate(
            torch.cat(
                [
                    time_feature,
                    freq_feature
                ],
                dim=-1
            )
        )

        fused = (
            gate * time_feature
            + (1.0 - gate)
            * freq_feature
        )

        fused = (
            fused + moe_feature
        )

        pooled = fused.mean(
            dim=1
        )

        pred = self.head(
            pooled
        ).squeeze(-1)

        return pred, {
            "balance_loss": balance_loss,
            "router": router_weights,
            "attention": attention
        }


print("所有模型组件定义完成")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# =========================================================
# 1. Padé activation
# =========================================================
class PadeActivation(nn.Module):
    def __init__(self, channels, m=3, n=2):
        super().__init__()

        self.m = m
        self.n = n

        self.a = nn.Parameter(
            torch.zeros(channels, m + 1)
        )

        self.b = nn.Parameter(
            torch.zeros(channels, n)
        )

        # 初始近似 identity
        with torch.no_grad():
            self.a[:, 0] = 0.0
            self.a[:, 1] = 1.0

            if m >= 2:
                self.a[:, 2:] = 0.0

            self.b[:] = 0.0

    def forward(self, x):
        # x: [B,C,L]

        z = torch.clamp(x, -5.0, 5.0)

        numerator = torch.zeros_like(z)

        for i in range(self.m + 1):
            coef = self.a[:, i].view(1, -1, 1)
            numerator = numerator + coef * z.pow(i)

        denominator_poly = torch.zeros_like(z)

        for j in range(1, self.n + 1):
            coef = self.b[:, j - 1].view(1, -1, 1)
            denominator_poly = (
                denominator_poly
                + coef * z.pow(j)
            )

        denominator = (
            1.0
            + torch.abs(denominator_poly)
        )

        return numerator / denominator


# =========================================================
# 2. Padé feature extractor
# =========================================================
class PadeFeatureExtractor(nn.Module):
    def __init__(
        self,
        in_channels=2,
        hidden_channels=16,
        dropout=0.1
    ):
        super().__init__()

        self.input_conv = nn.Conv1d(
            in_channels,
            hidden_channels,
            kernel_size=7,
            padding=3
        )

        self.pade1 = PadeActivation(
            hidden_channels
        )

        self.depthwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=5,
            padding=2,
            groups=hidden_channels
        )

        self.pointwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=1
        )

        self.pade2 = PadeActivation(
            hidden_channels
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.output_conv = nn.Conv1d(
            hidden_channels,
            in_channels,
            kernel_size=1
        )

        self.scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):
        # x: [B,L,2]

        residual = x

        y = x.transpose(1, 2)

        y = self.input_conv(y)
        y = self.pade1(y)

        y = self.depthwise(y)
        y = self.pointwise(y)

        y = self.pade2(y)
        y = self.dropout(y)

        y = self.output_conv(y)

        y = y.transpose(1, 2)

        return residual + self.scale * y


# =========================================================
# 3. PatchEmbedding
# =========================================================
class PatchEmbedding(nn.Module):
    def __init__(
        self,
        in_ch=2,
        patch_len=64,
        stride=64,
        d_model=64
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        self.proj = nn.Linear(
            in_ch * patch_len,
            d_model
        )

    def forward(self, x):
        # [B,L,C]

        p = x.unfold(
            dimension=1,
            size=self.patch_len,
            step=self.stride
        )

        p = p.contiguous().view(
            x.size(0),
            p.size(1),
            -1
        )

        return self.proj(p)


# =========================================================
# 4. Baseline MHA
# =========================================================
class MHATokenMixer(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        dropout=0.1
    ):
        super().__init__()

        self.attn = nn.MultiheadAttention(
            d_model,
            n_heads,
            dropout=dropout,
            batch_first=True
        )

    def forward(self, x):

        y, attn = self.attn(
            x,
            x,
            x,
            need_weights=False
        )

        return y, attn


# =========================================================
# 5. Small GDN-2
# =========================================================
class SmallGDN2(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads=4,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.dh = d_model // n_heads

        self.qkv = nn.Linear(
            d_model,
            3 * d_model,
            bias=False
        )

        self.erase_gate = nn.Linear(
            d_model,
            d_model
        )

        self.write_gate = nn.Linear(
            d_model,
            d_model
        )

        self.decay_gate = nn.Linear(
            d_model,
            n_heads
        )

        self.out_gate = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.norm = nn.LayerNorm(
            self.dh
        )

        self.dropout = nn.Dropout(
            dropout
        )

        # gate initialization
        nn.init.constant_(
            self.decay_gate.bias,
            2.0
        )

        nn.init.constant_(
            self.erase_gate.bias,
            -1.0
        )

        nn.init.constant_(
            self.write_gate.bias,
            -1.0
        )

        # learnable residual scale
        self.res_scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):

        B, N, D = x.shape

        H = self.n_heads
        dh = self.dh

        q, k, v = self.qkv(x).chunk(
            3,
            dim=-1
        )

        q = q.view(
            B, N, H, dh
        ).transpose(1, 2)

        k = k.view(
            B, N, H, dh
        ).transpose(1, 2)

        v = v.view(
            B, N, H, dh
        ).transpose(1, 2)

        q = F.normalize(q, dim=-1)
        k = F.normalize(k, dim=-1)

        erase = torch.sigmoid(
            self.erase_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        write = torch.sigmoid(
            self.write_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        decay = torch.sigmoid(
            self.decay_gate(x)
        ).transpose(1, 2)

        state = x.new_zeros(
            B,
            H,
            dh,
            dh
        )

        outputs = []

        for t in range(N):

            qt = q[:, :, t]
            kt = k[:, :, t]
            vt = v[:, :, t]

            et = erase[:, :, t]
            wt = write[:, :, t]

            dt = decay[
                :, :, t
            ].unsqueeze(-1).unsqueeze(-1)

            # decay
            state = dt * state

            # erase
            erase_vec = torch.einsum(
                "bhd,bhdv->bhv",
                et * kt,
                state
            )

            state = (
                state
                - kt.unsqueeze(-1)
                * erase_vec.unsqueeze(-2)
            )

            # write
            state = (
                state
                + kt.unsqueeze(-1)
                * (wt * vt).unsqueeze(-2)
            )

            # read
            yt = torch.einsum(
                "bhd,bhdv->bhv",
                qt,
                state
            )

            outputs.append(yt)

        y = torch.stack(
            outputs,
            dim=2
        )

        y = self.norm(y)

        y = (
            y.transpose(1, 2)
            .contiguous()
            .view(B, N, D)
        )

        gate = torch.sigmoid(
            self.out_gate(x)
        )

        y = self.out_proj(
            y * gate
        )

        y = self.dropout(y)

        y = self.res_scale * y

        return y, None


# =========================================================
# 6. Frequency branch
# =========================================================
class FrequencyBranch(nn.Module):
    def __init__(self, d_model):
        super().__init__()

        self.proj = nn.Linear(
            2 * d_model,
            d_model
        )

        self.ffn = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(
                d_model,
                2 * d_model
            ),
            nn.GELU(),
            nn.Linear(
                2 * d_model,
                d_model
            )
        )

    def forward(self, tokens):

        z = torch.fft.fft(
            tokens,
            dim=1,
            norm="ortho"
        )

        z = torch.cat(
            [
                z.real,
                z.imag
            ],
            dim=-1
        )

        z = self.proj(z)

        return z + self.ffn(z)


# =========================================================
# 7. Router
# =========================================================
class PrototypeRouter(nn.Module):
    def __init__(
        self,
        d_model,
        n_experts=4,
        top_k=2,
        temperature=0.5
    ):
        super().__init__()

        self.n_experts = n_experts
        self.top_k = top_k
        self.temperature = temperature

        self.prototypes = nn.Parameter(
            torch.randn(
                n_experts,
                d_model
            ) * 0.02
        )

    def forward(self, x):

        # 用 float32 避免 AMP overflow
        xf = x.float()
        pf = self.prototypes.float()

        xn = F.normalize(
            xf,
            dim=-1
        )

        pn = F.normalize(
            pf,
            dim=-1
        )

        logits = torch.einsum(
            "bnd,kd->bnk",
            xn,
            pn
        )

        logits = (
            logits
            / self.temperature
        )

        if self.top_k < self.n_experts:

            values, indices = torch.topk(
                logits,
                self.top_k,
                dim=-1
            )

            sparse = torch.full_like(
                logits,
                torch.finfo(
                    logits.dtype
                ).min
            )

            sparse.scatter_(
                -1,
                indices,
                values
            )

            weights = torch.softmax(
                sparse,
                dim=-1
            )

        else:

            weights = torch.softmax(
                logits,
                dim=-1
            )

        usage = weights.mean(
            dim=(0, 1)
        )

        balance_loss = (
            (
                usage
                - 1.0 / self.n_experts
            ) ** 2
        ).mean()

        weights = weights.to(
            dtype=x.dtype
        )

        return weights, balance_loss


# =========================================================
# 8. MoE
# =========================================================
class PatternMoE(nn.Module):
    def __init__(
        self,
        d_model,
        d_ff,
        n_experts
    ):
        super().__init__()

        self.experts = nn.ModuleList([
            nn.Sequential(
                nn.LayerNorm(d_model),

                nn.Linear(
                    d_model,
                    d_ff
                ),

                nn.GELU(),

                nn.Linear(
                    d_ff,
                    d_model
                )
            )
            for _ in range(n_experts)
        ])

    def forward(
        self,
        x,
        weights
    ):

        outputs = torch.stack(
            [
                expert(x)
                for expert in self.experts
            ],
            dim=-2
        )

        return (
            outputs
            * weights.unsqueeze(-1)
        ).sum(dim=-2)


# =========================================================
# 9. 完整 TFPSBearing
# =========================================================
class TFPSBearing(nn.Module):
    def __init__(
        self,
        mixer="mha",
        use_pade=False,
        seq_len=2048,
        patch_len=64,
        stride=64,
        d_model=64,
        n_heads=4,
        n_experts=4,
        top_k=2,
        d_ff=128,
        dropout=0.1
    ):
        super().__init__()

        self.mixer_name = mixer
        self.use_pade = use_pade

        # Padé
        if use_pade:
            self.pade_feature = (
                PadeFeatureExtractor(
                    in_channels=2,
                    hidden_channels=16,
                    dropout=dropout
                )
            )

        # patch
        self.patch = PatchEmbedding(
            in_ch=2,
            patch_len=patch_len,
            stride=stride,
            d_model=d_model
        )

        self.pre_norm = nn.LayerNorm(
            d_model
        )

        # time branch
        if mixer == "mha":

            self.time_mixer = (
                MHATokenMixer(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        elif mixer == "gdn2":

            self.time_mixer = (
                SmallGDN2(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        else:

            raise ValueError(
                f"Unknown mixer: {mixer}"
            )

        self.time_ffn = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                2 * d_model
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                2 * d_model,
                d_model
            )
        )

        # frequency
        self.freq = FrequencyBranch(
            d_model
        )

        # router
        self.router = PrototypeRouter(
            d_model,
            n_experts,
            top_k
        )

        # MoE
        self.moe = PatternMoE(
            d_model,
            d_ff,
            n_experts
        )

        # fusion
        self.fuse_gate = nn.Sequential(
            nn.Linear(
                2 * d_model,
                d_model
            ),
            nn.Sigmoid()
        )

        # prediction head
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                d_model // 2
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                d_model // 2,
                1
            )
        )

    def forward(self, x):

        # ===============================
        # Padé
        # ===============================
        if self.use_pade:
            x = self.pade_feature(x)

        # ===============================
        # Patch
        # ===============================
        tokens = self.patch(x)

        tokens = self.pre_norm(
            tokens
        )

        # ===============================
        # Time branch
        # ===============================
        time_mix, attention = (
            self.time_mixer(tokens)
        )

        time_feature = (
            tokens + time_mix
        )

        time_feature = (
            time_feature
            + self.time_ffn(
                time_feature
            )
        )

        # ===============================
        # Frequency branch
        # ===============================
        freq_feature = self.freq(
            tokens
        )

        # ===============================
        # Router
        # ===============================
        common = 0.5 * (
            time_feature
            + freq_feature
        )

        router_weights, balance_loss = (
            self.router(common)
        )

        # ===============================
        # MoE
        # ===============================
        moe_feature = self.moe(
            common,
            router_weights
        )

        # ===============================
        # Fusion
        # ===============================
        gate = self.fuse_gate(
            torch.cat(
                [
                    time_feature,
                    freq_feature
                ],
                dim=-1
            )
        )

        fused = (
            gate * time_feature
            + (1.0 - gate)
            * freq_feature
        )

        fused = (
            fused + moe_feature
        )

        pooled = fused.mean(
            dim=1
        )

        pred = self.head(
            pooled
        ).squeeze(-1)

        return pred, {
            "balance_loss": balance_loss,
            "router": router_weights,
            "attention": attention
        }


print("所有模型组件定义完成")
```

```python
import time
import copy
import numpy as np
import torch
import torch.nn.functional as F

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# =========================================================
# 参数量
# =========================================================
def count_params(model):
    return sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )


# =========================================================
# 验证 / 测试
# =========================================================
@torch.no_grad()
def evaluate(model, loader):

    model.eval()

    y_true_all = []
    y_pred_all = []

    for xb, yb in loader:

        xb = xb.to(
            DEVICE,
            non_blocking=True
        )

        yb = yb.to(
            DEVICE,
            non_blocking=True
        )

        pred, _ = model(xb)

        y_true_all.append(
            yb.detach().cpu().numpy()
        )

        y_pred_all.append(
            pred.detach().cpu().numpy()
        )

    y_true = np.concatenate(
        y_true_all
    )

    y_pred = np.concatenate(
        y_pred_all
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    metrics = {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2)
    }

    return metrics, y_true, y_pred


# =========================================================
# 训练
# =========================================================
def train_one(
    model,
    train_loader,
    val_loader,
    epochs=20
):

    model = model.to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    history = {
        "train_loss": [],
        "val_rmse": [],
        "val_mae": []
    }

    best_rmse = float("inf")
    best_state = None

    bad_epochs = 0

    use_amp = (
        DEVICE.type == "cuda"
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=use_amp
    )

    for epoch in range(
        1,
        epochs + 1
    ):

        model.train()

        epoch_losses = []

        for xb, yb in train_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=True
            )

            yb = yb.to(
                DEVICE,
                non_blocking=True
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            with torch.amp.autocast(
                device_type="cuda",
                enabled=use_amp
            ):

                pred, aux = model(xb)

                pred_loss = F.smooth_l1_loss(pred, yb, beta=0.1)



                balance_loss = aux.get(
                    "balance_loss",
                    torch.tensor(
                        0.0,
                        device=DEVICE
                    )
                )

                loss = (
                    pred_loss
                    + 0.01 * balance_loss
                )

            scaler.scale(
                loss
            ).backward()

            scaler.unscale_(
                optimizer
            )

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                max_norm=1.0
            )

            scaler.step(
                optimizer
            )

            scaler.update()

            epoch_losses.append(
                loss.item()
            )

        # validation
        val_metrics, _, _ = evaluate(
            model,
            val_loader
        )

        train_loss = float(
            np.mean(epoch_losses)
        )

        history[
            "train_loss"
        ].append(train_loss)

        history[
            "val_rmse"
        ].append(
            val_metrics["RMSE"]
        )

        history[
            "val_mae"
        ].append(
            val_metrics["MAE"]
        )

        print(
            f"Epoch {epoch:02d}/{epochs} | "
            f"train_loss={train_loss:.6f} | "
            f"val_MAE={val_metrics['MAE']:.6f} | "
            f"val_RMSE={val_metrics['RMSE']:.6f} | "
            f"val_R2={val_metrics['R2']:.4f}"
        )

        # early stopping
        if (
            val_metrics["RMSE"]
            < best_rmse
        ):

            best_rmse = (
                val_metrics["RMSE"]
            )

            best_state = copy.deepcopy(
                model.state_dict()
            )

            bad_epochs = 0

        else:

            bad_epochs += 1

            if bad_epochs >= PATIENCE:

                print(
                    f"Early stopping at epoch {epoch}"
                )

                break

    # 恢复最佳模型
    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    return model, history


# =========================================================
# 推理速度
# =========================================================
@torch.no_grad()
def inference_ms_per_sample(
    model,
    loader,
    warmup=3,
    runs=10
):

    model.eval()

    xb, _ = next(
        iter(loader)
    )

    xb = xb.to(DEVICE)

    # warmup
    for _ in range(warmup):
        model(xb)

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    start = time.perf_counter()

    for _ in range(runs):
        model(xb)

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()

    elapsed = (
        time.perf_counter()
        - start
    )

    ms_per_sample = (
        elapsed
        * 1000
        / (runs * xb.size(0))
    )

    return float(
        ms_per_sample
    )


print("训练辅助函数定义成功")

print(
    "count_params:",
    "OK"
    if "count_params" in globals()
    else "ERROR"
)

print(
    "train_one:",
    "OK"
    if "train_one" in globals()
    else "ERROR"
)

print(
    "evaluate:",
    "OK"
    if "evaluate" in globals()
    else "ERROR"
)
```

```python
class PrototypeRouter(nn.Module):

    def __init__(
        self,
        d_model,
        n_experts=4,
        top_k=2,
        temperature=0.5
    ):
        super().__init__()

        self.n_experts = n_experts
        self.top_k = top_k
        self.temperature = temperature

        self.prototypes = nn.Parameter(
            torch.randn(n_experts, d_model) * 0.02
        )

    def forward(self, x):

        # 路由强制用 float32，避免 AMP/FP16 溢出
        x_float = x.float()
        p_float = self.prototypes.float()

        x_norm = F.normalize(x_float, dim=-1)
        p_norm = F.normalize(p_float, dim=-1)

        logits = torch.einsum(
            "bnd,kd->bnk",
            x_norm,
            p_norm
        )

        logits = logits / self.temperature

        if self.top_k < self.n_experts:

            values, indices = torch.topk(
                logits,
                self.top_k,
                dim=-1
            )

            sparse = torch.full_like(
                logits,
                torch.finfo(logits.dtype).min
            )

            sparse.scatter_(
                -1,
                indices,
                values
            )

            weights = torch.softmax(
                sparse,
                dim=-1
            )

        else:

            weights = torch.softmax(
                logits,
                dim=-1
            )

        usage = weights.mean(dim=(0, 1))

        balance_loss = (
            (
                usage
                - 1.0 / self.n_experts
            ) ** 2
        ).mean()

        # 与主网络 dtype 保持一致
        weights = weights.to(x.dtype)

        return weights, balance_loss


print("PrototypeRouter 修复完成")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# =========================================================
# 1. Padé activation
# =========================================================
class PadeActivation(nn.Module):
    def __init__(self, channels, m=3, n=2):
        super().__init__()

        self.m = m
        self.n = n

        self.a = nn.Parameter(
            torch.zeros(channels, m + 1)
        )

        self.b = nn.Parameter(
            torch.zeros(channels, n)
        )

        # 初始近似 identity
        with torch.no_grad():
            self.a[:, 0] = 0.0
            self.a[:, 1] = 1.0

            if m >= 2:
                self.a[:, 2:] = 0.0

            self.b[:] = 0.0

    def forward(self, x):
        # x: [B,C,L]

        z = torch.clamp(x, -5.0, 5.0)

        numerator = torch.zeros_like(z)

        for i in range(self.m + 1):
            coef = self.a[:, i].view(1, -1, 1)
            numerator = numerator + coef * z.pow(i)

        denominator_poly = torch.zeros_like(z)

        for j in range(1, self.n + 1):
            coef = self.b[:, j - 1].view(1, -1, 1)
            denominator_poly = (
                denominator_poly
                + coef * z.pow(j)
            )

        denominator = (
            1.0
            + torch.abs(denominator_poly)
        )

        return numerator / denominator


# =========================================================
# 2. Padé feature extractor
# =========================================================
class PadeFeatureExtractor(nn.Module):
    def __init__(
        self,
        in_channels=2,
        hidden_channels=16,
        dropout=0.1
    ):
        super().__init__()

        self.input_conv = nn.Conv1d(
            in_channels,
            hidden_channels,
            kernel_size=7,
            padding=3
        )

        self.pade1 = PadeActivation(
            hidden_channels
        )

        self.depthwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=5,
            padding=2,
            groups=hidden_channels
        )

        self.pointwise = nn.Conv1d(
            hidden_channels,
            hidden_channels,
            kernel_size=1
        )

        self.pade2 = PadeActivation(
            hidden_channels
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.output_conv = nn.Conv1d(
            hidden_channels,
            in_channels,
            kernel_size=1
        )

        self.scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):
        # x: [B,L,2]

        residual = x

        y = x.transpose(1, 2)

        y = self.input_conv(y)
        y = self.pade1(y)

        y = self.depthwise(y)
        y = self.pointwise(y)

        y = self.pade2(y)
        y = self.dropout(y)

        y = self.output_conv(y)

        y = y.transpose(1, 2)

        return residual + self.scale * y


# =========================================================
# 3. PatchEmbedding
# =========================================================
class PatchEmbedding(nn.Module):
    def __init__(
        self,
        in_ch=2,
        patch_len=64,
        stride=64,
        d_model=64
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        self.proj = nn.Linear(
            in_ch * patch_len,
            d_model
        )

    def forward(self, x):
        # [B,L,C]

        p = x.unfold(
            dimension=1,
            size=self.patch_len,
            step=self.stride
        )

        p = p.contiguous().view(
            x.size(0),
            p.size(1),
            -1
        )

        return self.proj(p)


# =========================================================
# 4. Baseline MHA
# =========================================================
class MHATokenMixer(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        dropout=0.1
    ):
        super().__init__()

        self.attn = nn.MultiheadAttention(
            d_model,
            n_heads,
            dropout=dropout,
            batch_first=True
        )

    def forward(self, x):

        y, attn = self.attn(
            x,
            x,
            x,
            need_weights=False
        )

        return y, attn


# =========================================================
# 5. Small GDN-2
# =========================================================
class SmallGDN2(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads=4,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.dh = d_model // n_heads

        self.qkv = nn.Linear(
            d_model,
            3 * d_model,
            bias=False
        )

        self.erase_gate = nn.Linear(
            d_model,
            d_model
        )

        self.write_gate = nn.Linear(
            d_model,
            d_model
        )

        self.decay_gate = nn.Linear(
            d_model,
            n_heads
        )

        self.out_gate = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.norm = nn.LayerNorm(
            self.dh
        )

        self.dropout = nn.Dropout(
            dropout
        )

        # gate initialization
        nn.init.constant_(
            self.decay_gate.bias,
            2.0
        )

        nn.init.constant_(
            self.erase_gate.bias,
            -1.0
        )

        nn.init.constant_(
            self.write_gate.bias,
            -1.0
        )

        # learnable residual scale
        self.res_scale = nn.Parameter(
            torch.tensor(0.1)
        )

    def forward(self, x):

        B, N, D = x.shape

        H = self.n_heads
        dh = self.dh

        q, k, v = self.qkv(x).chunk(
            3,
            dim=-1
        )

        q = q.view(
            B, N, H, dh
        ).transpose(1, 2)

        k = k.view(
            B, N, H, dh
        ).transpose(1, 2)

        v = v.view(
            B, N, H, dh
        ).transpose(1, 2)

        q = F.normalize(q, dim=-1)
        k = F.normalize(k, dim=-1)

        erase = torch.sigmoid(
            self.erase_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        write = torch.sigmoid(
            self.write_gate(x)
        ).view(
            B, N, H, dh
        ).transpose(1, 2)

        decay = torch.sigmoid(
            self.decay_gate(x)
        ).transpose(1, 2)

        state = x.new_zeros(
            B,
            H,
            dh,
            dh
        )

        outputs = []

        for t in range(N):

            qt = q[:, :, t]
            kt = k[:, :, t]
            vt = v[:, :, t]

            et = erase[:, :, t]
            wt = write[:, :, t]

            dt = decay[
                :, :, t
            ].unsqueeze(-1).unsqueeze(-1)

            # decay
            state = dt * state

            # erase
            erase_vec = torch.einsum(
                "bhd,bhdv->bhv",
                et * kt,
                state
            )

            state = (
                state
                - kt.unsqueeze(-1)
                * erase_vec.unsqueeze(-2)
            )

            # write
            state = (
                state
                + kt.unsqueeze(-1)
                * (wt * vt).unsqueeze(-2)
            )

            # read
            yt = torch.einsum(
                "bhd,bhdv->bhv",
                qt,
                state
            )

            outputs.append(yt)

        y = torch.stack(
            outputs,
            dim=2
        )

        y = self.norm(y)

        y = (
            y.transpose(1, 2)
            .contiguous()
            .view(B, N, D)
        )

        gate = torch.sigmoid(
            self.out_gate(x)
        )

        y = self.out_proj(
            y * gate
        )

        y = self.dropout(y)

        y = self.res_scale * y

        return y, None


# =========================================================
# 6. Frequency branch
# =========================================================
class FrequencyBranch(nn.Module):
    def __init__(self, d_model):
        super().__init__()

        self.proj = nn.Linear(
            2 * d_model,
            d_model
        )

        self.ffn = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(
                d_model,
                2 * d_model
            ),
            nn.GELU(),
            nn.Linear(
                2 * d_model,
                d_model
            )
        )

    def forward(self, tokens):

        z = torch.fft.fft(
            tokens,
            dim=1,
            norm="ortho"
        )

        z = torch.cat(
            [
                z.real,
                z.imag
            ],
            dim=-1
        )

        z = self.proj(z)

        return z + self.ffn(z)


# =========================================================
# 7. Router
# =========================================================
class PrototypeRouter(nn.Module):
    def __init__(
        self,
        d_model,
        n_experts=4,
        top_k=2,
        temperature=0.5
    ):
        super().__init__()

        self.n_experts = n_experts
        self.top_k = top_k
        self.temperature = temperature

        self.prototypes = nn.Parameter(
            torch.randn(
                n_experts,
                d_model
            ) * 0.02
        )

    def forward(self, x):

        # 用 float32 避免 AMP overflow
        xf = x.float()
        pf = self.prototypes.float()

        xn = F.normalize(
            xf,
            dim=-1
        )

        pn = F.normalize(
            pf,
            dim=-1
        )

        logits = torch.einsum(
            "bnd,kd->bnk",
            xn,
            pn
        )

        logits = (
            logits
            / self.temperature
        )

        if self.top_k < self.n_experts:

            values, indices = torch.topk(
                logits,
                self.top_k,
                dim=-1
            )

            sparse = torch.full_like(
                logits,
                torch.finfo(
                    logits.dtype
                ).min
            )

            sparse.scatter_(
                -1,
                indices,
                values
            )

            weights = torch.softmax(
                sparse,
                dim=-1
            )

        else:

            weights = torch.softmax(
                logits,
                dim=-1
            )

        usage = weights.mean(
            dim=(0, 1)
        )

        balance_loss = (
            (
                usage
                - 1.0 / self.n_experts
            ) ** 2
        ).mean()

        weights = weights.to(
            dtype=x.dtype
        )

        return weights, balance_loss


# =========================================================
# 8. MoE
# =========================================================
class PatternMoE(nn.Module):
    def __init__(
        self,
        d_model,
        d_ff,
        n_experts
    ):
        super().__init__()

        self.experts = nn.ModuleList([
            nn.Sequential(
                nn.LayerNorm(d_model),

                nn.Linear(
                    d_model,
                    d_ff
                ),

                nn.GELU(),

                nn.Linear(
                    d_ff,
                    d_model
                )
            )
            for _ in range(n_experts)
        ])

    def forward(
        self,
        x,
        weights
    ):

        outputs = torch.stack(
            [
                expert(x)
                for expert in self.experts
            ],
            dim=-2
        )

        return (
            outputs
            * weights.unsqueeze(-1)
        ).sum(dim=-2)


# =========================================================
# 9. 完整 TFPSBearing
# =========================================================
class TFPSBearing(nn.Module):
    def __init__(
        self,
        mixer="mha",
        use_pade=False,
        seq_len=2048,
        patch_len=64,
        stride=64,
        d_model=64,
        n_heads=4,
        n_experts=4,
        top_k=2,
        d_ff=128,
        dropout=0.1
    ):
        super().__init__()

        self.mixer_name = mixer
        self.use_pade = use_pade

        # Padé
        if use_pade:
            self.pade_feature = (
                PadeFeatureExtractor(
                    in_channels=2,
                    hidden_channels=16,
                    dropout=dropout
                )
            )

        # patch
        self.patch = PatchEmbedding(
            in_ch=2,
            patch_len=patch_len,
            stride=stride,
            d_model=d_model
        )

        self.pre_norm = nn.LayerNorm(
            d_model
        )

        # time branch
        if mixer == "mha":

            self.time_mixer = (
                MHATokenMixer(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        elif mixer == "gdn2":

            self.time_mixer = (
                SmallGDN2(
                    d_model,
                    n_heads,
                    dropout
                )
            )

        else:

            raise ValueError(
                f"Unknown mixer: {mixer}"
            )

        self.time_ffn = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                2 * d_model
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                2 * d_model,
                d_model
            )
        )

        # frequency
        self.freq = FrequencyBranch(
            d_model
        )

        # router
        self.router = PrototypeRouter(
            d_model,
            n_experts,
            top_k
        )

        # MoE
        self.moe = PatternMoE(
            d_model,
            d_ff,
            n_experts
        )

        # fusion
        self.fuse_gate = nn.Sequential(
            nn.Linear(
                2 * d_model,
                d_model
            ),
            nn.Sigmoid()
        )

        # prediction head
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),

            nn.Linear(
                d_model,
                d_model // 2
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                d_model // 2,
                1
            )
        )

    def forward(self, x):

        # ===============================
        # Padé
        # ===============================
        if self.use_pade:
            x = self.pade_feature(x)

        # ===============================
        # Patch
        # ===============================
        tokens = self.patch(x)

        tokens = self.pre_norm(
            tokens
        )

        # ===============================
        # Time branch
        # ===============================
        time_mix, attention = (
            self.time_mixer(tokens)
        )

        time_feature = (
            tokens + time_mix
        )

        time_feature = (
            time_feature
            + self.time_ffn(
                time_feature
            )
        )

        # ===============================
        # Frequency branch
        # ===============================
        freq_feature = self.freq(
            tokens
        )

        # ===============================
        # Router
        # ===============================
        common = 0.5 * (
            time_feature
            + freq_feature
        )

        router_weights, balance_loss = (
            self.router(common)
        )

        # ===============================
        # MoE
        # ===============================
        moe_feature = self.moe(
            common,
            router_weights
        )

        # ===============================
        # Fusion
        # ===============================
        gate = self.fuse_gate(
            torch.cat(
                [
                    time_feature,
                    freq_feature
                ],
                dim=-1
            )
        )

        fused = (
            gate * time_feature
            + (1.0 - gate)
            * freq_feature
        )

        fused = (
            fused + moe_feature
        )

        pooled = fused.mean(
            dim=1
        )

        pred = self.head(
            pooled
        ).squeeze(-1)

        return pred, {
            "balance_loss": balance_loss,
            "router": router_weights,
            "attention": attention
        }


print("所有模型组件定义完成")
```

```python
import random
import numpy as np
import torch

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    # 尽量保证可复现
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

print("seed_everything 定义成功")
```

```python
import gc
import numpy as np
import pandas as pd
import torch

# =========================================================
# 1. 5 个随机种子
# =========================================================
SEEDS = [0, 1, 2, 3, 4]


# =========================================================
# 2. 四组实验
# =========================================================
EXPERIMENTS = [
    {
        "name": "TFPS-MHA",
        "mixer": "mha",
        "pade": False
    },
    {
        "name": "TFPS-GDN2",
        "mixer": "gdn2",
        "pade": False
    },
    {
        "name": "TFPS-Pade-MHA",
        "mixer": "mha",
        "pade": True
    },
    {
        "name": "TFPS-Pade-GDN2",
        "mixer": "gdn2",
        "pade": True
    }
]


# =========================================================
# 3. 保存所有结果
# =========================================================
all_results = []


for seed in SEEDS:

    print("\n")
    print("#" * 90)
    print(f"随机种子 SEED = {seed}")
    print("#" * 90)

    for cfg in EXPERIMENTS:

        print("\n" + "=" * 80)
        print(
            f"Seed={seed} | "
            f"开始训练: {cfg['name']}"
        )

        # -----------------------------------------
        # 清理上一个模型
        # -----------------------------------------
        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        # -----------------------------------------
        # 固定随机种子
        # -----------------------------------------
        seed_everything(seed)

        # -----------------------------------------
        # 创建模型
        # -----------------------------------------
        model = TFPSBearing(
            mixer=cfg["mixer"],
            use_pade=cfg["pade"],
            seq_len=SEQ_LEN,
            patch_len=PATCH_LEN,
            stride=PATCH_STRIDE,
            d_model=D_MODEL,
            n_heads=N_HEADS,
            n_experts=N_EXPERTS,
            top_k=TOP_K,
            d_ff=D_FF,
            dropout=DROPOUT
        )

        params = count_params(model)

        print(
            "参数量:",
            f"{params:,}"
        )

        # -----------------------------------------
        # 训练
        # -----------------------------------------
        model, hist = train_one(
            model,
            train_loader,
            val_loader,
            epochs=EPOCHS
        )

        # -----------------------------------------
        # Test
        # -----------------------------------------
        metrics, y_true, y_pred = evaluate(
            model,
            test_loader
        )

        # -----------------------------------------
        # 推理速度
        # -----------------------------------------
        speed = inference_ms_per_sample(
            model,
            test_loader
        )

        # -----------------------------------------
        # GDN residual scale
        # -----------------------------------------
        gdn_scale = np.nan

        if cfg["mixer"] == "gdn2":

            gdn_scale = float(
                model.time_mixer
                .res_scale
                .detach()
                .cpu()
                .item()
            )

        # -----------------------------------------
        # Padé scale
        # -----------------------------------------
        pade_scale = np.nan

        if cfg["pade"]:

            pade_scale = float(
                model.pade_feature
                .scale
                .detach()
                .cpu()
                .item()
            )

        # -----------------------------------------
        # 保存结果
        # -----------------------------------------
        row = {
            "seed": seed,
            "model": cfg["name"],

            "MAE": metrics["MAE"],
            "RMSE": metrics["RMSE"],
            "R2": metrics["R2"],

            "Params": params,
            "ms_per_sample": speed,

            "GDN_scale": gdn_scale,
            "Pade_scale": pade_scale
        }

        all_results.append(row)

        print("\n测试结果:")
        print(row)

        # 防止 GPU 内存累计
        del model

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()


# =========================================================
# 4. 每一次实验结果
# =========================================================
all_results_df = pd.DataFrame(
    all_results
)

print("\n")
print("=" * 90)
print("全部 5-seed 实验结果")
print("=" * 90)

display(
    all_results_df
)


# 保存
all_results_df.to_csv(
    "/content/tfps_pade_gdn2_5seeds_raw.csv",
    index=False
)
```
